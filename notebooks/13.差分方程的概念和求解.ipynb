### 4.1 差分方程的概念

#### 4.1.1 差分

微分方程是自变量连续取值的问题，但在很多实际问题中，有些变量不是连续取值的。例如经济变量收入、储蓄等都是时间序列，自变量 $t$ 取值为 0, 1, 2, ...，数学上把这种变量称为离散型变量。通常用差商来描述因变量对自变量的变化速度。设函数 $x_k = x(k)$，$k$ 只取非负整数，称该变量 $x_{k+1}-x_k$ 为函数 $x_k$ 的 **差分**，也称为函数 $x_k$ 的一阶差分，记为 $Δ_{x_k}$，即 $Δ_{x_k} = x_{k+1}-x_k$。一阶差分的差分称为二阶差分 $Δ^2_{x_k}$，即 $Δ^2_{x_k} = Δ(Δ_{x_k}) = Δ_{x_{k+1}}-Δ_{x_k} = (x_{k+2}-x_{k+1})-(x_{k+1}-x_k)$。类似地，可定义三阶差分、四阶差分......

#### 4.1.2 差分方程

含有未知函数 $x_k$ 的差分的方程称为差分方程。若差分方程中所含未知函数即未知函数的各阶差分均为一次的，则称该差分方程为 **线性差分方程**。n阶常系数线性差分方程的一般形式为：

$
x_{k+n}+a_1x_{k+n-1}+a_2x_{k+n-2}+\ldots+a_nx_k = f(k) \quad (1)
$

其中，$a_1,a_2,\ldots,a_n$ 是常数，且 $a_n≠0$。上述方程对应的齐次方程为：

$
x_{k+n}+a_1x_{k+n-1}+a_2x_{k+n-2}+\ldots+a_nx_k = 0 \quad (2)
$

#### 4.1.3 齐次差分方程的解

对于 n 阶常系数线性齐次差分方程 (2)，它的特征方程为：

$
λ^n+a_1λ^{n-1}+\ldots+a_n = 0 \quad (3)
$

根据特征根的不同情况，求齐次方程 (2) 的通解。
- 若特征方程 (3) 有 n 个互不相同的实根 $λ_1,λ_2,\ldots,λ_n$，则对应的齐次差分方程的通解为：$c_1λ_1^k+c_2λ_2^k+\ldots+c_nλ_n^k \quad (c_1,c_2,\ldots,c_n 为任意常数)$
- 若 $λ$ 是特征方程 (3) 的 m 重实根，通解中对应于 $λ$ 的项为 $(c_1+c_2n+\ldots+c_mn^{m-1})λ^k$。
- 若特征方程 (3) 有单重复根 $λ = α+βi$，通解中对应的项为 $c_1ρ^k\cos(φk)+c_2ρ^k\sin(φk)$，其中，$ρ = \sqrt{α^2+β^2}$ 为 $λ$ 的模，$φ = \arctan\frac{β}{α}$ 为 $λ$ 的幅角，$c_1,c_2$ 为任意常数。
- 若 $λ = α+βi$ 是特征方程 (3) 的 m 重复根，则通解中对应的项为 $(c_1+c_2n+\ldots+c_mn^{m-1})ρ^k\cos(φk)+(c_{m+1}+c_{m+2}n+\ldots+c_{2m}n^{m-1})ρ^k\sin(φk)$，其中，$ρ = \sqrt{α^2+β^2}$，$φ = \arctan\frac{β}{α}$，$c_i(i=1,2,\ldots,2m)$ 为任意常数。

#### 4.1.4 非齐次差分方程的解

非齐次差分方程的解 = 齐次差分方程的通解 ($x_k$) + 非齐次差分方程的一个特解 ($x^*_k$)。对于特殊形式的 $f(k)$ 可以使用待定系数法求非齐次方程的特解。例如当 $f(k) = b^kp_m(k)$，$p_m(k)$ 为 k 的 m 次多项式时，可以证明：
- 若 b 不是特征根，则非齐次方程 (1) 有形如 $b^kq_m(k)$ 的特解，$q_m(k)$ 也是 k 的 m 次多项式。
- 若 b 是 r 重特征根，则非齐次方程 (1) 有形如 $b^kk^rq_m(k)$ 的特解。

进而可利用待定系数法求出 $q_m(k)$，从而得到非齐次方程 (1) 的一个特解。

### 4.2 差分方程求解

#### 4.2.1 常系数线性齐次差分方程求解

**例1**. 求斐波那契数列的解
$
\begin{cases}
F_n = F_{n-1}+F_{n-2} \\
F_1 = F_2 = 1
\end{cases}
$

`sympy` 是用于符号运算的 Python 库，下面是一个使用 `sympy` 求解斐波那契数列的例子：

```python
from sympy import Function, rsolve, symbols

# 定义符号
n = symbols('n')
# 定义函数
F = Function('F')
# 求解差分方程
f = F(n) - F(n-1) - F(n-2)
solution = rsolve(f, F(n), {F(1): 1, F(2): 1})
print(solution)
```


````python
from sympy import Function, rsolve
from sympy.abc import n
# 定义函数y和符号n
y = Function('y')
# 定义通项公式
f = y(n)-y(n-1)-y(n-2)
# 定义初始条件
y0 = {y(1):1,y(2):1}
# 求解递推式
res = rsolve(f, y(n),y0)
print('齐次差分方程的通解为\nF(n)=',res)
````

在这个例子中，我们定义了递推式$f(n+2) = f(n-1) + f(n-2)$，初始条件为$f(0) =1，f(1) =1$。

然后使用rsolve()函数求解递推式，得到递推式的通解：

$$f(n) = \sqrt{5}*{(\frac{1}{2} + \frac{\sqrt{5}}{2})}^n/5 - \sqrt{5}{( -\frac{\sqrt{5}}{2} + \frac{1}{2})}^n/5$$

#### 4.2.2 常系数线性非齐次差分方程求解

**例2**.求差分方程$x_{k+1}-x_k = 3+2k$的通解

````python
from sympy.abc import k
from sympy import simplify,var
# 定义函数y
y = var('y',cls=Function)
# 定义通项公式
f = y(k+1)-y(k)-3-2*k
# 求解递推式（无初始值）
f1 = rsolve(f, y(k))
print('非齐次差分方程的通解为\nxk=',f1)
# simplify()表达式化简
f2 = simplify(f1)
print('非齐次差分方程的通解为（展开式）\nxk=',f2)
````

#### 4.2.3 差分方程的递推解法

差分方程有些可以表示为递推形式，由已知数据，只需按照递推形式即可求解。

**例3（贷款问题）**.某人从银行贷款购房，若他今年初贷款100万元，月利率0.5%，每月还款8000元，建立差分方程计算他每年末欠银行多少钱，多长时间能还清贷款。

记第$k$个月末欠款总额为$x_k$（单位：元），$r$为月利率$r$ = 0.5%，$y$为当月还款额度$y = 8000$（单位：元），则第k+1个月末欠银行的钱为$x_{k+1}$（单位：元）

差分方程为：

$$
\begin{cases}{x_{k+1} = x_k（1+r）-y}\\
{x_0 = 1000000}
\end{cases}

$$

````python
# 定义初始贷款，月利率
x0 = 1000000
r = 0.5/100
# 月还款额度 
N = 500
n = 0
y = 8000
# x1为第一个月欠款
x1 = x0*(1+r)-y
while n<=N and x1>0:
    n+=1;
    # 每12个月打印一次剩余欠款
    if n%12==0: 
        print("第%d个月末欠款：x(%d)=%.d（元）"%(n,n,x1))
    # 重新赋值
    x0 = x1
    x1 = x0*(1+r)-y

print("还款月数n=",n+1)
year = (n+1)//12
month = n+1-12*((n+1)//12)
print("还款%d年%d个月"%(year,month))
````

````python
# 求每个月剩余欠款通项公式

# 定义函数x
x = Function('x')
# 定义通项公式
f = x(k+1)-x(k)*(1+r)+8000
# 定义初始条件
x0 = {x(0):1000000}
# 求解递推式
res1 = rsolve(f, x(k),x0)
print('第k个月末欠银行的钱为\nx(k)=',res1)
````

### 4.3 差分方程数值求解

> 来源：教材参考实现《数学建模算法与应用》15.3 节的练习稿（MIT，Zhanghan Wang），代码保持原样，仅按需改写数据路径并清空输出。

````python
# 作n阶差分折线图
import matplotlib.pyplot as plt

def diff_plot(y, n):
    if y.shape[0] < n:
        print('Error')
        return
    for i in range(n):
        y = np.array([y[i]-y[i-1] if i != 0 else y[0] for i in range(y.shape[0])])
    plt.figure()
    plt.ylim(y.mean()-(y.max()-y.min())*3, (y.max()-y.min())*3+y.max())
    plt.plot(np.linspace(0, y.shape[0]-1, y.shape[0]), y)
    plt.title('{} Order Difference of Data'.format(n))
    plt.show()
````

````python
# 例15.6

import numpy as np

y0 = np.array([11, 16, 25, 12, 12, 18, 26, 14, 13, 20, 27, 15, 15, 24, 30, 15, 16, 25, 32, 17])
diff_plot(y0, 1)
y = y0[8: 20]
x = np.array([y0[4:16], y0[0:12], np.ones(12)]).T
z = np.matmul(np.linalg.inv(np.matmul(x.T, x)), np.matmul(x.T, y))
print(z)

pred_y = y0.copy()
for i in range(10):
    pred_y = np.r_[pred_y, [z[0] * pred_y[i-4] + z[1] * pred_y[i-8] + z[2]]]
    
print(pred_y)
````
