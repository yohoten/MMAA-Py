# 31 线性规划

## 1. 问题背景

线性规划在一组线性约束下优化线性目标，是资源分配、生产计划与运输问题的基础模型。本节用 `scipy.optimize.linprog` 求解标准形式，并对照教材算例核对结果。

## 2. 学习目标

- 掌握 `linprog` 的参数含义与不等式方向
- 把实际问题整理为 A_ub·x ≤ b_ub 的标准形
- 用约束回代与目标值校验求解结果

## 3. 环境与数据

- 依赖：scipy
- 数据：无（内置示例）

## 4. 方法与实现

### 4.1 linprog 标准形求解

```python
import scipy.optimize

# document: https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.linprog.html#scipy.optimize.linprog
# function name: scipy.optimize.linprog
# function: linear programming.
# 例1.2的程序
c = [2, 3, -5]
A_ub = [[-2, 5, -1],
        [1, 3, 1]]
b_ub = [-10, 12]
A_eq = [[1, 1, 1]]
b_eq = [7]
bounds = [(0, None), (0, None), (0, None)]
res = scipy.optimize.linprog(c=[-x for x in c], A_ub=A_ub, b_ub=b_ub, A_eq=A_eq, b_eq=b_eq, bounds=bounds)
print(res)
print(res.get('x'))
```

下面的算例把教材中的线性规划模型写成 `linprog` 的标准形：最大化目标取负后转为最小化，约束统一为 `A_ub @ x <= b_ub`。

```python
import numpy as np
from scipy.optimize import linprog

# max 2x + 3y  ->  min -2x - 3y
c = [-2, -3]
A_ub = [[1, 2], [4, 0], [0, 4]]
b_ub = [8, 16, 12]
bounds = [(0, None), (0, None)]

res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method="highs")
print(res.status, res.message)
print("x, y =", np.round(res.x, 4))
print("最大目标值 =", round(-res.fun, 4))
```

## 5. 小结

`linprog` 默认求最小值，约束需整理成 ≤ 形式；求解后应回代检查可行性，并留意无界与不可行两种状态。
