## 3. 多元线性回归模型

### 3.1 问题的提出

现实生活中引起被解释变量变化的因素并非仅只一个解释变量，可能有多个解释变量。例如，产出往往受各种投入要素（如资本、劳动、技术等）的影响；销售额往往受价格和公司对广告费的投入影响等。因此，在一元线性模型的基础上，提出多元线性回归模型，即解释变量个数≥2。

### 3.2 一般形式

- 定义：设随机变量 $y$ 与一般变量 $x_1, x_2, \ldots, x_p$ 的线性回归模型为：

$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_p x_p + \varepsilon \tag{1}

$$

### 3.3 矩阵表示法

若有 $n$ 组观测数据 $(x_{i1}, x_{i2}, \ldots, x_{ip}; y_i)$ (i=1,2,...,n)，则多元线性回归模型可表示为：

$$
\begin{cases}
y_1 = \beta_0 + \beta_1 x_{11} + \beta_2 x_{12} + \cdots + \beta_p x_{1p} + \varepsilon_1 \\
y_2 = \beta_0 + \beta_1 x_{21} + \beta_2 x_{22} + \cdots + \beta_p x_{2p} + \varepsilon_2 \\
\vdots \\
y_n = \beta_0 + \beta_1 x_{n1} + \beta_2 x_{n2} + \cdots + \beta_p x_{np} + \varepsilon_n
\end{cases} \tag{2}

$$

(2)式写成矩阵形式为：

$$
Y = X \beta + \varepsilon

$$
其中，

$$
Y = \begin{pmatrix} y_1 \\ y_2 \\ \vdots \\ y_n \end{pmatrix}, \quad X = \begin{pmatrix} 1 & x_{11} & x_{12} & \cdots & x_{1p} \\ 1 & x_{21} & x_{22} & \cdots & x_{2p} \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ 1 & x_{n1} & x_{n2} & \cdots & x_{np} \end{pmatrix}, \quad \beta = \begin{pmatrix} \beta_0 \\ \beta_1 \\ \vdots \\ \beta_p \end{pmatrix}, \quad \varepsilon = \begin{pmatrix} \varepsilon_1 \\ \varepsilon_2 \\ \vdots \\ \varepsilon_n \end{pmatrix}

$$

### 3.4 基本假定

1. **零均值假定**：假定随机干扰项 $\varepsilon$ 的期望向量为 0：

$$
E(\varepsilon_i) = 0 \quad (i = 1, 2, \ldots, n)

$$
2. **正态性假设**：每个观测值的误差项 $\varepsilon$ 服从正态分布：

$$
\varepsilon_i \sim N(0, \sigma^2)

$$
3. **独立性假设**：每个观测值的误差项 $\varepsilon$ 是相互独立的：

$$
Cov(\varepsilon_i, \varepsilon_j) = 0 \quad (i \neq j)

$$
4. **同方差性假设**：

$$
Var(\varepsilon_i) = \sigma^2

$$
5. **无多重共线性假设**：自变量之间不存在高度相关性。

### 3.5 参数估计

用样本统计量 $\widehat{\beta}_0, \widehat{\beta}_1, \ldots, \widehat{\beta}_p$ 估计回归方程中的参数 $B_0, B_1, \ldots, B_p$，一般形式为：使因变量的观测值与估计值之间的离差平方和达到最小，求得 $B_0, B_1, \ldots, B_p$。即

$$
Q(\widehat{\beta}_0, \widehat{\beta}_1, \widehat{\beta}_2, \ldots, \widehat{\beta}_p) = \sum_{i=1}^{n} (y_i - \widehat{y_i})^2 = \sum_{i=1}^{n} \varepsilon_i^2

$$
使 $Q$ 达到最小。

### 3.6 参数估计：矩阵形式

$$
\rho(\widehat{\beta}) = (Y - X\widehat{\beta})^T(Y - X\widehat{\beta})

$$
求解回归参数，对上式关于 $\beta$ 求偏导，并令其为零：

$$
\frac{\partial Q(\beta)}{\partial \beta} = -2X^T(Y - X\beta) = 0 \\
\Rightarrow (X^TX)\widehat{\beta} = X^TY

$$
当矩阵 $X$ 列满秩时，$X^TX$ 为可逆方阵，此时：

$$
\widehat{\beta} = (X^TX)^{-1}X^TY

$$


## 4. 多元线性回归方程的检验

### 4.1 拟合优度检验：调整后的判定系数

调整后的判定系数为：

$$
\overline{R}^2 = 1 - \frac{(1 - R^2) (n - 1)}{n - p - 1}

$$
其中，$n$ 表示样本个数，$p$ 表示模型中自变量的个数。

### 4.2 拟合优度检验：估计标准误差

估计标准误差是回归方程残差的均方根：

$$
s_e = \sqrt{\frac{\sum (y_i - \widehat{y_i})^2}{n - p - 1}} = \sqrt{MSE}

$$
$s_e$ 反映了实际观测值 $y$ 与回归估计值之间的差异程度。

### 4.3 回归方程的显著性检验：F 检验

目的：检验自变量与因变量之间的线性关系是否显著。

1. 提出假设：

$$
H_0: \beta_1 = \beta_2 = \cdots = \beta_p = 0 \quad $(回归方程线性关系不显著)$

$$

2. 构造 $F$ 统计量：

$$
F = \frac{SSR/p}{SSE/(n - p - 1)} \sim F(p, n - p - 1)

$$

3. 计算 $F$ 统计量的值和伴随概率 $p$。
4. 统计决策（显著性水平 $\alpha = 0.05$）：
- 如果 $p < \alpha$，拒绝 $H_0$，即回归方程线性关系存在。
- 反之，不拒绝 $H_0$。

回归系数的显著性检验：t检验
目的：检验自变量对因变量的线性影响是否显著。

1. 提出假设：

$$
H_{0}^{(i)}: \beta_{i} = 0 \quad (i=0,1,\ldots,p) \quad $(解释变量对被解释变量的影响不显著)$

$$

2. 构造 $t$ 统计量：

$$
t_{i} = \frac{\hat{\beta}_{i}/\sqrt{c_{ii}}}{\sqrt{SSE/(n - p - 1)}} \sim t(n - p - 1)

$$
其中 $c_{ii}$ 是 $(X^TX)^{-1}$ 中的第 $(i, i)$ 元素。

3. 计算 $t$ 统计量的值和伴随概率 $p$。
4. 统计决策（显著性水平 $\alpha = 0.05$），如果 $p < \alpha$，拒绝 $H_{0}$，即系数显著。反之，不拒绝 $H_{0}$。

### 4.4 多元回归方程的预测

1. **点估计**：对于给定的 $x_1^{(0)}, x_2^{(0)}, \ldots, x_p^{(0)}$ 代入回归方程，得到：

$$
\hat{y}_{0} = \hat{\beta}_{0} + \hat{\beta}_{1} x_{1}^{(0)} + \hat{\beta}_{2} x_{2}^{(0)} + \cdots + \hat{\beta}_{p} x_{p}^{(0)}

$$
用 $\hat{y}_{0}$ 作为 $y$ 在点 $(x_{1}^{(0)}, x_{2}^{(0)}, \ldots, x_{p}^{(0)})$ 的预测值。

2. **区间估计**：记 $s = \sqrt{SSE/(n - p - 1)}, x_{0} = [1, x_{1}^{(0)}, x_{2}^{(0)}, \ldots, x_{p}^{(0)}]$，则 $y_{0}$ 的置信度为 $\alpha$ 的预测区间为：

$$
\hat{y}_{0} \pm t_{\alpha/2}(n - p - 1) s \sqrt{1 + x_{0}^{T}(X^TX)^{-1}x_{0}}

$$


## 5. 多重共线性

### 5.1 什么是多重共线性？

多重共线性（multicollinearity）是指在多元回归分析中，自变量之间存在高度相关性或线性关系的情况。具体来说，当自变量之间存在较高的相关性时，可能会出现以下情况：
- **系数估计不准确**：多重共线性会导致回归模型中自变量的系数估计不准确，使得对自变量的解释和预测能力下降。
- **系数符号错误**：多重共线性还可能导致回归模型中自变量的系数符号与理论预期相反，这会使得解释变量对因变量的影响方向出现错误。
- **假设检验失效**：多重共线性会导致回归模型中的假设检验结果不可靠，使得无法对自变量的影响进行显著性检验。
- **不稳定的模型**：多重共线性可能导致回归模型的不稳定性，即微小的数据变动可能导致模型结果的较大变化。

### 5.2 多重共线性的识别

检测多重共线性最简单的一种方法是计算模型中各对自变量之间的相关系数，并对各相关系数进行显著性检验。

若有一个或多个相关系数显著，就表示模型中所用的自变量之间相关，存在多重共线性。假如出现下列情况，则暗示存在多重共线性：
- 模型中各对自变量之间显著相关
- 当模型的线性关系检验 (F检验) 显著时，集合所有回归系数的 $t$ 检验却不显著
- 回归系数的正负号与预期的相反


## 6. 线性回归模型的正则化

### 6.1 岭回归

当 $x_1, x_2, \ldots, x_p$ 之间存在多重共线性关系时，即 $|X^TX| \approx 0$。

$$
\widehat{\beta} = (X^TX)^{-1}X^TY

$$
此时 $X^TX$ 不可逆，为了保证回归系数的求解，岭回归模型在目标函数上加了一个 L2 范数的惩罚项：

$$
e(\beta) = \sum (y - X\beta)^2 + \lambda \sum \beta^2

$$
其中 $\lambda$ 表示正则化参数（$\lambda > 0$），此时

$$
\widehat{\beta}(\lambda) = (X^TX + \lambda I)^{-1}X^TY

$$
其中 $I$ 表示单位矩阵。单位矩阵的对角线上全是 1，这也是岭回归名称的由来。

### 6.2 岭回归：岭参数的选择

岭参数的选择有多种方法，主要有：
1. 岭迹法
2. 均方误差法

**1. 岭迹法**：观察岭迹曲线，原则上应该选择使 $\widehat{\beta}(\lambda)$ 稳定的最小 $\lambda$ 值。

**2. 均方误差法**：
岭估计的均方误差：

$$
MSE(\widehat{\beta}(\lambda)) = E|\tilde{\beta}(\lambda) - \beta|^2

$$
- 则取最小值处的作为岭参数。

### 6.3 LASSO 回归

LASSO 回归和岭回归方法类似，只不过 LASSO 是在目标函数上加了 L1 正则化：

$$
\rho(\beta) = \sum(Y - X\beta)^2 + \lambda \sum |\beta|

$$
其中 $\lambda$ 表示正则化参数（$\lambda > 0$）。

由于惩罚项是关于回归系数 $\beta$ 的绝对值之和，因此惩罚项在零点处不可导，最小二乘法在此失效，故采用坐标下降法。

与岭回归模型类似，$\lambda$ 值的确定可以通过定性的可视化方法。

### 6.4 岭回归和 LASSO 回归区别

岭回归与 LASSO 回归最大的区别在于目标函数加入的惩罚项：岭回归引入的是 L2 范数惩罚项，LASSO 回归引入的是 L1 范数惩罚项。

随着正则化强度的增大，LASSO 的取值会逐渐变小，L1 正则化会将参数压缩到 0，而 L2 正则化只会让参数尽量小，不会取到 0。所以在 L1 正则化逐渐增强的过程中，相对不重要的特征的参数会比相对重要的特征的参数更快地变成 0。由此，L1 正则化本质上是一个特征选择的过程，选出少量但重要的特征，以防止过拟合问题。

而 L2 正则化在增强过程中，会尽量让每个特征对模型都有一些贡献，相对不重要的特征的参数会非常接近 0。
