# 38 偏最小二乘回归

## 1. 问题背景

当自变量多、彼此共线且样本量小时，偏最小二乘回归通过提取潜变量同时解释自变量与因变量，常用于光谱与化学计量数据。

## 2. 学习目标

- 理解潜变量提取的基本思想
- 会用交叉验证确定成分数
- 能解读回归系数与变量重要性

## 3. 环境与数据

- 依赖：scikit-learn
- 数据：无（内置示例）

## 4. 方法与实现

### 4.1 偏最小二乘回归分析

[参考网站](https://www.cnblogs.com/duye/p/9031511.html）
- 当数据量小，甚至比变量维数还小，而相关性又比较大时使用，这个方法甚至优于主成分回归。  

[sklearn.cross_decomposition.PLSRegression文档](https://scikit-learn.org/stable/modules/generated/sklearn.cross_decomposition.PLSRegression.html)

```python
from sklearn.cross_decomposition import PLSRegression
X = [[0., 0., 1.], [1.,0.,0.], [2.,2.,2.], [2.,5.,4.]]
Y = [[0.1, -0.2], [0.9, 1.1], [6.2, 5.9], [11.9, 12.3]]
pls2 = PLSRegression(n_components=2)
pls2.fit(X, Y)
Y_pred = pls2.predict(X)
print(Y_pred)
```

## 5. 小结

成分数过少欠拟合、过多过拟合；应以交叉验证误差最小为准，而不是一味增加成分。
